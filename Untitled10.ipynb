```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Đọc dữ liệu từ file CSV
df = pd.read_csv('/content/drive/MyDrive/TMG/vihallu-train.csv')

# Loại bỏ các dòng có giá trị thiếu (nếu có)
df.dropna(inplace=True)

# Kết hợp 3 cột văn bản thành 1 cột input duy nhất
# Dùng ký tự đặc biệt [SEP] để mô hình hiểu ranh giới giữa các phần
df['text_input'] = "Ngữ cảnh: " + df['context'] + " [SEP] " + \
                   "Câu hỏi: " + df['prompt'] + " [SEP] " + \
                   "Trả lời: " + df['response']

# Chuyển đổi nhãn (label) từ dạng chữ sang dạng số (ví dụ: 'extrinsic' -> 0, 'intrinsic' -> 1)
label_encoder = LabelEncoder()
df['label_encoded'] = label_encoder.fit_transform(df['label'])

# In ra để xem thử 5 dòng đầu và số lượng mỗi nhãn
print("5 dòng dữ liệu đầu tiên sau khi xử lý:")
print(df.head())
print("\nSố lượng mẫu cho mỗi nhãn:")
print(df['label'].value_counts())

# Phân chia dữ liệu thành tập train và validation (80% train, 20% validation)
train_texts, val_texts, train_labels, val_labels = train_test_split(
    df['text_input'].tolist(),
    df['label_encoded'].tolist(),
    test_size=0.2,
    random_state=42,
    stratify=df['label_encoded'] # Giữ tỷ lệ các nhãn ở cả 2 tập
)

print(f"\nKích thước tập huấn luyện: {len(train_texts)}")
print(f"Kích thước tập kiểm định: {len(val_texts)}")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from sklearn.metrics import f1_score, accuracy_score, precision_recall_fscore_support

# Tên model PhoBERT trên Hugging Face Hub
MODEL_NAME = "vinai/phobert-base"

# Tải Tokenizer để mã hóa văn bản
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Tải Model đã được pre-train và cấu hình cho bài toán phân loại
# num_labels là số lượng nhãn độc nhất trong dữ liệu của bạn
num_labels = len(df['label_encoded'].unique())
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=num_labels)

# Đảm bảo mô hình chạy trên GPU nếu có
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print(f"Sử dụng thiết bị: {device}")
print(f"Số lượng nhãn: {num_labels}")
```

```python
# Mã hóa (tokenize) dữ liệu văn bản
# max_length=256 là một lựa chọn hợp lý, bạn có thể tăng lên nếu câu của bạn dài hơn
train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=256)
val_encodings = tokenizer(val_texts, truncation=True, padding=True, max_length=256)

# Tạo một lớp Dataset tùy chỉnh
class HallucinationDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)

# Tạo đối tượng Dataset cho tập train và validation
train_dataset = HallucinationDataset(train_encodings, train_labels)
val_dataset = HallucinationDataset(val_encodings, val_labels)
```

```python
# Định nghĩa hàm để tính toán các chỉ số đánh giá
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='macro')
    acc = accuracy_score(labels, preds)
    return {
        'accuracy': acc,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }

# Thiết lập các tham số cho quá trình huấn luyện
training_args = TrainingArguments(
    output_dir='./results',
    learning_rate=2e-5,              # Thư mục để lưu kết quả và model checkpoint
    num_train_epochs=3,              # Số lần lặp qua toàn bộ tập dữ liệu (3-5 epochs là hợp lý)
    per_device_train_batch_size=16,  # Kích thước lô cho mỗi thiết bị khi huấn luyện
    per_device_eval_batch_size=32,   # Kích thước lô cho mỗi thiết bị khi đánh giá
    warmup_steps=500,                # Số bước khởi động cho learning rate
    weight_decay=0.01,               # Hệ số suy giảm trọng số
    logging_dir='./logs',            # Thư mục để lưu log
    logging_steps=10,
    evaluation_strategy="epoch",     # Đánh giá mô hình sau mỗi epoch
    save_strategy="epoch",           # Lưu mô hình sau mỗi epoch
    load_best_model_at_end=True,     # Tải mô hình tốt nhất sau khi huấn luyện xong
    metric_for_best_model="f1",      # Chọn F1-score làm tiêu chí để chọn mô hình tốt nhất
)

# Khởi tạo đối tượng Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics
)

# Bắt đầu huấn luyện!
print("Bắt đầu quá trình huấn luyện...")
trainer.train()
```

```python
# Đánh giá hiệu suất của mô hình trên tập validation
print("\nĐánh giá mô hình trên tập kiểm định:")
eval_results = trainer.evaluate()

print(f"\nKết quả đánh giá:")
for key, value in eval_results.items():
    print(f"  {key}: {value:.4f}")

# Kết quả sẽ có dạng như:
# eval_loss: 0.XXXX
# eval_accuracy: 0.XXXX
# eval_f1: 0.XXXX  <-- Đây là chỉ số quan trọng nhất bạn cần theo dõi
# eval_precision: 0.XXXX
# eval_recall: 0.XXXX
```
